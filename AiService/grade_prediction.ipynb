```python
import pandas as pd
```

```python
data = {
    "Moisture": [12.5, 15.0, 18.5, 13.0, 16.2],
    "Foreign_Matter": [1.2, 2.5, 3.0, 1.5, 2.0],
    "Damaged_Grains": [2.0, 3.5, 4.0, 2.5, 3.0]
}

df = pd.DataFrame(data)

df
```

```python
def predict_grade(row):

    if row["Moisture"] < 14 and row["Foreign_Matter"] < 2 and row["Damaged_Grains"] < 3:
        return "A"

    elif row["Moisture"] <= 17:
        return "B"

    else:
        return "C"
```

```python
df["Predicted_Grade"] = df.apply(predict_grade, axis=1)

df
```
